```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=["the glass of milk","the glass of juice","the cup of tea","I am a good boy","I am a good developer","You are a good developer"]

```

```python
## Define the vocabulary size
voc_size=10000
```

```python
one_hot_repr=[one_hot(words,voc_size)for words in sent]
print(one_hot_repr)  ## This returns the index of the words in the vocabulary
```

```python
## Word Embedding system(To convert those indexes)

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential

## to train the RNN using the sentences we need to create the matrix of the sentences and for all the sentences we need to have same number of words
## So we need to pad the sentences
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(embedded_docs)
print(embedded_docs.shape)
```

```python
## Feature representation
dim=10
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
