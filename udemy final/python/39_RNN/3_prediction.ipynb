```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence ## Explanation below in the markdown
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense

```

```python
from tensorflow.keras.models import load_model
```

```python
word_index=imdb.get_word_index()
reverse_word_index={value:key for key, value in word_index.items()}

## Reverse word index means at which index we have what words
```

```python
model=load_model('model_imdb.h5')
```

```python
model.get_weights() 
## What Does It Actually Return?

##It returns:A list of weight matrices and bias vectors for every layer in order.Each layer contributes its parameters.
```

```python
## 2 helper functions


## Func to decode reviews
def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i-3,'?') for i in encoded_review])  ## based on the OHE we match the indexes and decode

## Func to process user inp
def preprocess_text(text):
    words=text.lower().split()
    encoded_review=[word_index.get(word,2)+3 for word in words]
    padded_review=sequence.pad_sequences([encoded_review],maxlen=500)
    return padded_review
```

```python
def predict_sentiment(review):
    preprocessed_input = preprocess_text(review)
    prediction = model.predict(preprocessed_input)

    sentiment = 'positive' if prediction[0][0] > 0.5 else 'negative'
    return sentiment,prediction[0][0]
```

Class 0 → Negative

Class 1 → Neutral

Class 2 → Positive

For one sample:

[[0.1, 0.7, 0.2]]

This means:

10% probability → class 0  ------->prediction[0][0]

70% probability → class 1  ------->prediction[1][0]

20% probability → class 2  ------->prediction[2][0]

All values sum to 1.

As we are using only binary classification so we return prediction[0][0]

```python
example='THis movie was fantastic! The acting was great and the plot was thrilling'

sentiment,score=predict_sentiment(example)
print(f'The review {example}')
print(f'Predicted sentiment: {sentiment}')
print(f'Score: {score}')
```
