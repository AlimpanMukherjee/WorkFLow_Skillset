```python
import numpy as np

class PositionalEncoding:
    
    def __init__(self, d_model=6, max_sequence_length=10):
        """
        d_model: dimension of model
        max_sequence_length: maximum number of tokens
        """
        self.d_model = d_model
        self.max_sequence_length = max_sequence_length
        
        # Create positional encoding once during initialization
        self.PE = self._create_positional_encoding()
    
    
    def _create_positional_encoding(self):
        """
        Creates positional encoding matrix of shape:
        (max_sequence_length, d_model)
        """
        
        position = np.arange(self.max_sequence_length)[:, np.newaxis]
        # (max_seq_len, 1)

        div_term = np.exp(
            np.arange(0, self.d_model, 2) * (-np.log(10000.0) / self.d_model)
        )
        # (d_model/2,)

        PE = np.zeros((self.max_sequence_length, self.d_model))

        PE[:, 0::2] = np.sin(position * div_term)
        PE[:, 1::2] = np.cos(position * div_term)

        return PE
    
    
    def forward(self):
        """
        Returns positional encoding matrix
        """
        return self.PE
```

```python
pe = PositionalEncoding(d_model=6, max_sequence_length=10)

pos_encoding = pe.forward()

print("Positional Encoding shape:", pos_encoding.shape)
print(pos_encoding)
```
